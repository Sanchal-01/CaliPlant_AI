```python
import os

# California ke kuch top plants ki list (Didi apke ke liye initial list)
plants_to_track = [
    'California_Poppy', 
    'Poison_Oak', 
    'Joshua_Tree', 
    'Coast_Live_Oak', 
    'California_Sagebrush'
]

base_path = '../03_Data/train'

for plant in plants_to_track:
    path = os.path.join(base_path, plant)
    os.makedirs(path, exist_ok=True)
    print(f"Directory ready: {plant}")

print("Folders ban gaye hain! Ab isme images bharne ki bari hai.")
```

```python
import os

base_path_train = '../03_Data/train'
base_path_val = '../03_Data/validation'

# 1. Train folder se saare folders ki list uthana (Dynamic)
all_plants = [d for d in os.listdir(base_path_train) if os.path.isdir(os.path.join(base_path_train, d))]

print(f" Total {len(all_plants)} plants detected in Train folder.")

# 2. Validation folder mein missing folders banao
for plant in all_plants:
    val_path = os.path.join(base_path_val, plant)
    if not os.path.exists(val_path):
        os.makedirs(val_path)
        print(f" Created missing validation folder: {plant}")

print("\n Structure Check Complete: Train and Validation ab synced hain!")
```
